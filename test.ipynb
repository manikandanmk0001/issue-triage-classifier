```python
import pandas as pd

cats = {
    "model error": [32700, 33005, 34459, 34075, 33000, 33772, 34292, 34145,
                    34822, 34315, 34312, 34151, 34310, 33371, 34222, 33108,
                    32691, 34260, 34477, 34045, 33710, 32834, 34279],
    "label looks wrong": [33721, 34483, 34781, 34529, 33858, 34344, 34035, 33745],
    "no content": [33428],
    "ambiguous": [35014, 33830, 32807, 34170, 33904, 32961, 34532, 32946,
                  34615, 33093, 34139, 34232, 33045, 34203, 32725, 33214,
                  32914, 34286],
}
lookup = {i: c for c, ids in cats.items() for i in ids}

e = pd.read_csv("data/errors.csv")
e["category"] = e["id"].map(lookup)
assert e["category"].notna().all(), "some ids are missing a category"
e.to_csv("data/errors_categorized.csv", index=False)
print(e["category"].value_counts())
```
